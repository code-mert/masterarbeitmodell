# Diskussion:

Viele unterschiedliche, lange LLM-Formulierungen werden pauschal auf das übergeordnete "Ulkus" herabgestuft. Kritikpunkt: Jeder komplexe Satz des LLMs wird als exakter Treffer gewertet selbst wenn das LLM in den Nebensätzen falsche Details hinzugefügt hat.

Fazit & Qualitative Synthese: Kategorie „Wundtyp“
1. Charakteristik & Herausforderung der Kategorie
Die Bestimmung des Wundtyps ist die primäre ätiologische Weichenstellung in der Wunddokumentation. Die besondere Herausforderung bei einer rein bildbasierten Beurteilung liegt darin, dass der genaue Wundtyp klinisch häufig erst durch Anamnesedaten und apparative Diagnostik (z. B. Pulsstatus/ABI bei Ulcus cruris, Diabetes-Historie oder OP-Berichte) abgesichert wird. Ein reines Foto lässt die Ätiologie bei bestimmten Wunden (z. B. Unterschenkel- vs. Fußulzerationen) naturgemäß ambivalent.

2. Vergleich der 3 Prompting-Ansätze
Zero-Shot:
Zeigt eine solide Basiserkennung bei visuell eindeutigen Wundtypen.
Neigt bei Unklarheiten entweder zu sehr generischen Antworten (Ulkus, Chronische Wunde) oder zum Raten spezifischer Ätiologien ohne hinreichende visuelle Evidenz.
Few-Shot:
Führt zu einer deutlich verbesserten Schema-Compliance und konsistenteren Benennung gemäß dem vorgegebenen Katalog.
Verhindert das Abschweifen in zu freie Beschreibungen im Raw-Output.
Two-Stage (Chain-of-Thought):
Liefert im Raw-Text die klinisch tiefgründigste Analyse (z. B. expliziter Hinweis auf Stoma-Nähe bei postoperativer Wunddehiszenz oder Eviszeration).
Formulierte ausführliche Erklärungen enthalten oft klinische Hedging-Begriffe („Verdacht auf“, „bildbasiert wahrscheinlich“), die im Raw-Text wertvoll sind, aber ein präzises Normalisierungs-Mapping erfordern.
3. Stärken & Schwachstellen im Modellverhalten
✅ Wo waren die Modelle stark?
Klare visuelle Leitsymptome: Wundtypen mit prägnanten visuellen Merkmalen wurden sehr zuverlässig erkannt:
Verbrennungswunden (typisches Erscheinungsbild mit Blasenbildung, epidermaler Schädigung).
Postoperative Wunden (sichtbare chirurgische Nähte, Klammern, Stoma-Nähe oder dehisziente Narben).
Dekubitus (typische Prädilektionsstellen wie Ferse oder Os sacrum mit Druckzeichen/Nekrosen).
❌ Was war nicht so gut / Wo lagen die Grenzen?
Differenzierung bei Bein- und Fußulzerationen: Die Abgrenzung zwischen Ulcus cruris venosum, arteriosum, mixtum und diabetischem Fußulkus ohne klinische Zusatzdaten führt bei Modellen wie auch Experten zu Abweichungen.
Formale Trennung von Wundtyp und Spezifikation: Das LLM packte Freitext-Details (z. B. Gradangaben bei Verbrennungen oder den Begriff Platzbauch) häufig direkt in das Feld wundtyp, anstatt das dafür vorgesehene Feld wundtyp_spezifizierung zu nutzen.
4. Methodische Implikationen & Was man bedenken muss
Experten-Diskrepanz (Inter-Rater-Variabilität):
Auch die Ground-Truth-Daten von Experte 1 und Experte 2 zeigen bei Wundtyp Freitext-Variationen (z. B. Platzbauch vs. Postoperative Wunde abdominal). Das Modell kann daher nicht zwingend besser sein als die Konsistenz der Experten untereinander.
Nutzen des klinisch-semantischen Mappings:
Das Mapping von Freitexten (wie Platzbauch $\rightarrow$ Postoperative Wunde) ist essenziell, um sprachliche Unterschiede von echten Inhaltsfehlern zu trennen.
Empfehlung für zukünftige Prompts:
Im System-Prompt sollte strikt vorgegeben werden, dass im Feld wundtyp ausschließlich der übergeordnete Haupt-Wundtyp eingetragen wird, während Komplikationen (Platzbauch) und Gradangaben (2. Grad) konsequent in das Feld wundtyp_spezifizierung gehören.

Es gibt mehrere prägnante Fälle, in denen selbst die beiden Experten bei ein und demselben Bild zu völlig unterschiedlichen Wundtypen kommen.

Hier sind die prägnantesten Beispiele aus den Ground-Truth-Daten:

1. Fundamental konträre Haupt-Diagnosen
wunde_10:

Experte 1: arterielles Ulcus (Gefäßerkrankung / pAVK)
Experte 2: Dekubitus (Druckgeschwür am Vorfuß)
➔ Bedeutung: Das ist ein medizinisch grundlegender Unterschied in der Ätiologie. Experte 2 merkt sogar an: „Lokalisation nicht genau definierbar, für mich sieht es aus wie am Vorfuß“.
wunde_28:

Experte 1: Ulkus (unspezifisches Geschwür)
Experte 2: Dekubitus (Druckursache)
➔ Bedeutung: Während Experte 1 sich wegen fehlender Anamnese vorsichtig nur auf Ulkus festlegt, diagnostiziert Experte 2 einen Fersendekubitus.
2. Spezifische Ätiologie vs. Unscharfer Freitext
wunde_20 & wunde_26:

Experte 1: Ulkus
Experte 2: Diabetisches Fußulkus
➔ Bedeutung: Experte 2 schätzt das Wundbild am Fuß als typische Diabetiker-Komplikation ein, während Experte 1 mangels Blutzucker-/Patientendaten beim generischen Oberbegriff bleibt.
wunde_25:

Experte 1: Wunde am li Fußrücken mit Sehnenfreilegung (Deskriptive Wundbeschreibung)
Experte 2: Diabetisches Fußulkus (Spezifische Ätiologie)
wunde_09:

Experte 1: Nekrose (Gewebszustand)
Experte 2: Ulkus cruris arteriosum (Arterielles Unterschenkel-/Fußgeschwür)
Warum das für deine Arbeit & das LLM extrem wichtig ist:
Inter-Rater-Discrepancy: Diese Abweichungen zeigen, dass eine rein visuelle Wundtyp-Bestimmung ohne Anamnese selbst für menschliche Experten eine Herausforderung mit Ermessensspielraum darstellt.
Fairness gegenüber dem LLM: Wenn Experte 1 arterielles Ulcus sagt und Experte 2 Dekubitus, liegt das LLM bei fast jeder Wahl gegenüber einem der beiden Experten „falsch“ (rot/orange) – obwohl beide Diagnosen in der Praxis als Differenzialdiagnosen aufgestellt würden!
