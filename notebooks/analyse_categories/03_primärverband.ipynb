Key-Fact vorab:
Von den 60 Wundbildern erzielt das Modell bei 55 Wunden (91,7 %) mit mindestens einem Ansatz einen Treffer auf Produktfamilien-Ebene.

Es verbleiben genau 5 absolute Härtefälle (8,3 %) (wunde_01, wunde_08, wunde_09, wunde_22, wunde_25), bei denen kein einziger der 3 LLM-Ansätze (Zero-Shot, Few-Shot, Two-Stage) auch nur eine Übereinstimmung mit einem der Experten erzielen konnte.

Warum diese 5 Fälle fehlschlagen (Klinische Ursachen-Analyse):
1. Spezialtherapie Unterdruck-Wundtherapie (NPWT / Suprasorb CNP)
Betroffene Wunden: wunde_01, wunde_22
Ground Truth der Experten: Beide Experten fordern Suprasorb CNP (Vakuumpumpen-System / Negative Pressure Wound Therapy).
LLM-Empfehlungen: Schlägt Alginate und Superabsorber vor (Suprasorb A + Ag, Vliwasorb Pro).
Ursache: NPWT ist eine apparative Spezialversorgung. Aus einem statischen 2D-Foto erkennt das LLM zwar eine tiefe, exsudierende Wunde, schlägt aber fälschlicherweise eine hoch-absorbierende Wundauflage statt eines apparativen Unterdrucksystems vor.
2. Leere Ground Truth ("Kein Primärverband indiziert")
Betroffene Wunden: wunde_09 (und teilweise wunde_08)
Ground Truth der Experten: Beide Experten geben keinen Primärverband an (Feld bleibt leer / "").
LLM-Empfehlungen: Schlägt präventiv Wundauflagen vor (z. B. Suprasorb G, Suprasorb F, Suprasorb X).
Ursache: Das LLM ist im Prompt darauf konditioniert, stets eine Verbandsempfehlung abzugeben. Wenn ein Wundbild keine Abdeckung erfordert (z.B. primär abheilende Naht), neigt das Modell zu Über-Behandlung (Over-Prescription).
3. Rehydrierung / Hydrogel vs. Saugmedien
Betroffene Wunde: wunde_25
Ground Truth der Experten: Experten fordern Suprasorb G Gel-Kompresse oder amorphes Gel (Rehydrierung von trockenem Schorf).
LLM-Empfehlungen: Schlägt Suprasorb Liquacel Pro (Hydrofiber) und Suprasorb P (Schaumstoff) vor.
Ursache: Das Modell interpretiert Glanz oder Belag auf dem Foto fälschlicherweise als Exsudat und wählt absorbierende Schaumstoffe/Hydrofiber statt feuchtigkeitsspendendem Hydrogel.

```python
import category_loader

# Erstellt die Heatmap-Verwechslungsmatrix für Two-Stage (Best-of-Both Konsens)
category_loader.plot_primaerverband_confusion_matrix(model_approach="Two-Stage", evaluation_mode="best_of_both")
```

Die Zahlen aus den Daten bestätigen das zu 100 %:

Two-Stage: Schlägt in 49 von 60 Wundbildern (81,7 %) ein Suprasorb P-Produkt (Schaumstoff) vor!
Zero-Shot: Schlägt in 44 von 60 Wundbildern (73,3 %) Suprasorb P vor.
Zum Vergleich Experte 2: Nutzt ebenfalls in 45 von 60 Wundbildern (75,0 %) Suprasorb P.
Zum Vergleich Experte 1: Nutzt in 33 von 60 Wundbildern (55,0 %) Suprasorb P.
Warum wählt das Modell so extrem oft Suprasorb P?
1. Der "Allrounder"-Effekt in der Wundversorgung
Schaumstoffe (Suprasorb P) gelten in der klinischen Praxis als das universelle Standardmedium bei exsudierenden Wunden (z. B. Ulcus cruris, Dekubitus). Auch Experte 2 setzt bei 3/4 aller Wunden auf Schaumstoffe.

2. Risikominimierungs-Strategie des LLMs ("Safety First")
Das Modell lernt während des Trainings: Mit einem Schaumstoffverband macht man klinisch fast nie etwas gefährliches falsch (solange die Wunde nicht völlig austrocknet). Es dient dem LLM als sichere Fallback-Lösung, wenn der Befund uneindeutig ist.

3. Der Grund für den hohen 71,1 % Score
Diese starke Bevorzugung von Suprasorb P ist paradoxerweise der Hauptgrund dafür, warum Two-Stage so hohe F1-Scores (71,1 %) erzielt! Da auch die Experten sehr oft Schaumstoffe fordern, staubt das Modell durch den Suprasorb P-Focus extrem viele Treffer ab.

4. Die Schattenseite für deine Masterarbeit
Das Problem dabei ist: Das Modell ist etwas "faul" bei Spezialindikationen. Anstatt feine Nuancen zu erkennen (z. B. wenn nur ein dünnes Wunddistanzgitter wie Solvaline oder feuchtigkeitsspendende Hydrobalance wie Suprasorb X benötigt wird), greift es präventiv lieber gleich zum dicken Schaumstoff.
